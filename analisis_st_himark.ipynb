## Gráfico 2 — Matriz de priorización: shakemap vs reportes (dropdown por evento)

```python
# Matriz 2x2 con dropdown por evento, UMBRALES POR MEDIANA (recalculados por evento)

REGIMENES = {
    'CONFIRMADO':          {'color': '#b1000c', 'desc': 'Plan estándar'},
    'SORPRESA NEGATIVA':   {'color': '#8b5cf6', 'desc': 'Sumar recursos'},
    'SILENCIO SOSPECHOSO': {'color': '#ff7f0e', 'desc': 'Verificar in situ'},
    'TRANQUILO':           {'color': '#4a86a8', 'desc': 'Monitoreo pasivo'},
}

def regimen_de(esp, obs, th_x, th_y):
    alta_esp = esp > th_x
    alta_obs = obs > th_y
    if alta_esp and alta_obs: return 'CONFIRMADO'
    if alta_esp and not alta_obs: return 'SILENCIO SOSPECHOSO'
    if not alta_esp and alta_obs: return 'SORPRESA NEGATIVA'
    return 'TRANQUILO'

def datos_matriz(evento):
    """Calcula severidad observada y arma el dataframe con régimen.
    
    NOTA IMPORTANTE: la intensidad esperada (eje X) sale de SHAKE_OFICIAL,
    que se estimó visualmente del PNG del shakemap del sismo principal.
    Esa misma asignación se reutiliza para los 4 eventos del dropdown
    (limitación: no hay shakemaps específicos para las réplicas).
    Los umbrales se calculan como la MEDIANA de los valores observados
    en ese evento — son dinámicos y se recalculan al cambiar de evento.
    """
    t0, t1 = EVENTOS[evento]
    sub = df[(df['time'] >= t0) & (df['time'] <= t1)]
    sev = sub.groupby('location')[DAMAGE].mean().mean(axis=1)
    n_rep = sub.groupby('location').size()

    rows = []
    for i in range(1, 20):
        rows.append({
            'barrio': NEIGH[i],
            'esperada': SHAKE_OFICIAL[i],
            'observada': float(sev.get(i, 0)),
            'n_reportes': int(n_rep.get(i, 0)),
        })
    mx = pd.DataFrame(rows)

    # Umbrales por mediana del evento
    th_x = mx['esperada'].median()
    th_y = mx['observada'].median()

    mx['régimen'] = mx.apply(lambda r: regimen_de(r['esperada'], r['observada'], th_x, th_y), axis=1)
    return mx, th_x, th_y

def plot_matriz_evento(evento):
    mx, th_x, th_y = datos_matriz(evento)

    fig, ax = plt.subplots(figsize=(11, 8))

    # Sombras de cuadrantes
    x_min, x_max = 3, 7.5
    y_min, y_max = -0.5, mx['observada'].max() + 0.7
    ax.axvspan(th_x, x_max, ymin=(th_y - y_min)/(y_max - y_min), ymax=1,
               facecolor='#b1000c', alpha=.07)
    ax.axvspan(x_min, th_x, ymin=(th_y - y_min)/(y_max - y_min), ymax=1,
               facecolor='#8b5cf6', alpha=.07)
    ax.axvspan(th_x, x_max, ymin=0, ymax=(th_y - y_min)/(y_max - y_min),
               facecolor='#ff7f0e', alpha=.07)
    ax.axvspan(x_min, th_x, ymin=0, ymax=(th_y - y_min)/(y_max - y_min),
               facecolor='#4a86a8', alpha=.07)

    # Líneas de umbral
    ax.axvline(th_x, color='gray', lw=1, ls='--', alpha=.5)
    ax.axhline(th_y, color='gray', lw=1, ls='--', alpha=.5)

    # Puntos por barrio
    for _, r in mx.iterrows():
        color = REGIMENES[r['régimen']]['color']
        size = 50 + min(r['n_reportes'] / 30, 250)
        ax.scatter(r['esperada'], r['observada'], s=size,
                   color=color, edgecolor='black', linewidth=.7, alpha=.85, zorder=3)
        ax.annotate(r['barrio'], (r['esperada'], r['observada']),
                    xytext=(6, 4), textcoords='offset points', fontsize=8.5)

    # Etiquetas de cuadrantes en esquinas
    ax.text(0.98, 0.97, 'CONFIRMADO\n(plan estándar)', transform=ax.transAxes,
            fontsize=10, color='#b1000c', weight='bold', ha='right', va='top',
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', edgecolor='#b1000c', alpha=.85))
    ax.text(0.98, 0.03, 'SILENCIO SOSPECHOSO\n(verificar in situ)', transform=ax.transAxes,
            fontsize=10, color='#ff7f0e', weight='bold', ha='right', va='bottom',
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', edgecolor='#ff7f0e', alpha=.85))
    ax.text(0.02, 0.97, 'SORPRESA NEGATIVA\n(sumar recursos)', transform=ax.transAxes,
            fontsize=10, color='#8b5cf6', weight='bold', ha='left', va='top',
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', edgecolor='#8b5cf6', alpha=.85))
    ax.text(0.02, 0.03, 'TRANQUILO', transform=ax.transAxes,
            fontsize=10, color='#4a86a8', weight='bold', ha='left', va='bottom',
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', edgecolor='#4a86a8', alpha=.85))

    ax.set_xlabel('Intensidad esperada (shakemap, estimación visual)')
    ax.set_ylabel('Severidad reportada (promedio de daño material)')
    ax.set_title(f'Matriz de priorización — {evento}\n'
                 f'(umbrales por mediana: esp={th_x:.2f}, obs={th_y:.2f}; tamaño = n reportes)')
    ax.set_xlim(x_min, x_max)
    ax.set_ylim(y_min, y_max)
    for s in ['top', 'right']:
        ax.spines[s].set_visible(False)
    plt.tight_layout()
    return fig, mx

# Widget para cambiar de evento
evento_matriz_select = pn.widgets.Select(
    name='Evento sísmico',
    options=list(EVENTOS.keys()),
    value='Sismo principal (8/04)',
)

@pn.depends(evento_matriz_select.param.value)
def actualizar_matriz(evento):
    fig, _ = plot_matriz_evento(evento)
    plt.close(fig)
    return fig

pn.Column(evento_matriz_select, actualizar_matriz)
```

## Gráfico 3 — Paleta bivariada: severidad × dispersión por barrio (dropdown por categoría)

```python
# Paleta bivariada 3x3 estilo Stevens 
# Lectura de la paleta:
# - Filas: severidad. Fila 0 (arriba) = severidad ALTA, fila 2 (abajo) = severidad BAJA.
# - Columnas: dispersión sigma. Col 0 (izq) = sigma BAJA, col 2 (der) = sigma ALTA.
#
# Esquinas clave:
# - Top-left (sev alta + sigma baja) → ROJO INTENSO: situación crítica con datos confiables.
# - Top-right (sev alta + sigma alta) → MORADO: crítico pero con dispersión interna.
# - Bottom-left (sev baja + sigma baja) → GRIS CLARO: tranquilo y confiable.
# - Bottom-right (sev baja + sigma alta) → TURQUESA SUAVE: bajo nivel pero datos ruidosos.
PALETA_BIV = [
    ['#9d2f4d', '#a8559e', '#5b6c9c'],   # severidad alta
    ['#d5798a', '#cba0c1', '#92a4c4'],   # severidad media
    ['#e8e8e8', '#cce0d8', '#9bcfd0'],   # severidad baja
]
SEV_LABELS = ['Alta', 'Media', 'Baja']
SIG_LABELS = ['σ baja', 'σ media', 'σ alta']

def color_bivariado(sev, sigma, sev_max=8.0, sigma_max=3.5):
    if pd.isna(sev) or pd.isna(sigma):
        return '#cccccc'
    sev_norm = min(sev / sev_max, 1.0)
    sigma_norm = min(sigma / sigma_max, 1.0)
    sev_bin = 2 if sev_norm > 0.66 else (1 if sev_norm > 0.33 else 0)
    sigma_bin = 2 if sigma_norm > 0.66 else (1 if sigma_norm > 0.33 else 0)
    fila = 2 - sev_bin
    return PALETA_BIV[fila][sigma_bin]

# Datos del sismo principal
t0_main, t1_main = EVENTOS['Sismo principal (8/04)']
sub_main = df[(df['time'] >= t0_main) & (df['time'] <= t1_main)]

def plot_bivariado_matplotlib(categoria_es):
    """Renderiza el mapa con paleta bivariada usando matplotlib + geopandas."""
    cat = [c for c, e in CAT_ES.items() if e == categoria_es][0]
    sev_b = sub_main.groupby('location')[cat].mean()
    sig_b = sub_main.groupby('location')[cat].std()

    # Asignar color a cada barrio
    gdf_loc = gdf.copy()
    gdf_loc['sev'] = gdf_loc['location'].map(sev_b)
    gdf_loc['sigma'] = gdf_loc['location'].map(sig_b)
    gdf_loc['color'] = gdf_loc.apply(
        lambda r: color_bivariado(r['sev'], r['sigma']), axis=1)

    # Plot
    fig = plt.figure(figsize=(13, 7))
    gs = fig.add_gridspec(1, 2, width_ratios=[3, 1], wspace=0.02)
    ax_map = fig.add_subplot(gs[0])
    ax_leg = fig.add_subplot(gs[1])

    # Mapa
    gdf_loc.plot(color=gdf_loc['color'], edgecolor='black', linewidth=0.7, ax=ax_map)
    for _, row in gdf_loc.iterrows():
        c = row['geometry'].centroid
        ax_map.annotate(str(int(row['location'])), (c.x, c.y),
                        ha='center', va='center', fontsize=10,
                        color='black', weight='bold')
    ax_map.set_title(f'Severidad × Dispersión — {categoria_es}\n(sismo principal 8/04)')
    ax_map.set_axis_off()

    # Leyenda 3x3 a la derecha
    ax_leg.set_xlim(-0.3, 3.2)
    ax_leg.set_ylim(-1.0, 3.5)
    ax_leg.set_aspect('equal')
    for i in range(3):
        for j in range(3):
            rect = plt.Rectangle((j, 2 - i), 0.95, 0.95,
                                  facecolor=PALETA_BIV[i][j], edgecolor='white', linewidth=2)
            ax_leg.add_patch(rect)
    for i, lab in enumerate(SEV_LABELS):
        ax_leg.text(-0.08, 2 - i + 0.5, lab, ha='right', va='center', fontsize=11)
    for j, lab in enumerate(SIG_LABELS):
        ax_leg.text(j + 0.5, -0.15, lab, ha='center', va='top', fontsize=10)
    ax_leg.text(1.5, -0.78, 'Dispersión (σ)', ha='center', fontsize=12, weight='bold')
    ax_leg.set_axis_off()

    plt.tight_layout()

    # "Severidad" colocado con coordenadas reales del axes — no se superpone con las etiquetas
    pos = ax_leg.get_position()
    fig.text(pos.x0 - 0.035, pos.y0 + pos.height / 2,
             'Severidad', rotation=90, va='center', ha='center', fontsize=12, weight='bold')

    return fig

# Widget dropdown
cat_select_biv = pn.widgets.Select(
    name='Categoría',
    options=[CAT_ES[c] for c in CATEGORIES],
    value='Edificios',
)

@pn.depends(cat_select_biv.param.value)
def actualizar_bivariado(categoria):
    fig = plot_bivariado_matplotlib(categoria)
    plt.close(fig)
    return fig

pn.Column(cat_select_biv, actualizar_bivariado)
```

## Gráfico 4 — Distribución de los reportes por barrio (dropdown por evento)

```python
# Ridge plot con dropdown por evento
def ridge_plot_evento(evento):
    t0, t1 = EVENTOS[evento]
    sub = df[(df['time'] >= t0) & (df['time'] <= t1)].dropna(subset=['shake_intensity'])
    
    fig, ax = plt.subplots(figsize=(10, 8.5))
    x_grid = np.linspace(0, 10, 200)
    
    # Ordenar barrios por mediana descendente
    medians = sub.groupby('neighborhood')['shake_intensity'].median().sort_values(ascending=False)
    order = medians.index.tolist()
    
    offset = 0; gap = 0.9
    colors_rg = plt.cm.plasma(np.linspace(0.15, 0.85, len(order)))
    
    for i, neigh in enumerate(order):
        vals = sub.loc[sub['neighborhood']==neigh, 'shake_intensity'].values
        if len(vals) < 5:
            ax.text(-0.15, offset+.25, neigh, ha='right', va='center', fontsize=9, color='gray')
            ax.text(10.15, offset+.25, f'n={len(vals)}', ha='left', va='center', fontsize=8, color='gray')
            ax.hlines(offset+.25, 0, 10, colors='lightgray', linestyles=':', linewidth=.5)
            offset += 1
            continue
        try:
            kde = gaussian_kde(vals, bw_method=0.4)
            y = kde(x_grid)
            y = y / max(y.max(), 1e-6) * gap
        except Exception:
            offset += 1; continue
        
        ax.fill_between(x_grid, offset, offset+y, color=colors_rg[i], alpha=.78,
                        linewidth=.6, edgecolor='white')
        ax.plot(x_grid, offset+y, color='white', lw=.6)
        med = np.median(vals)
        ax.plot([med, med], [offset, offset+0.5], color='black', lw=1.2, alpha=.7)
        ax.text(-0.15, offset+.25, neigh, ha='right', va='center', fontsize=9)
        ax.text(10.15, offset+.25, f'n={len(vals):,}', ha='left', va='center', fontsize=8, color='gray')
        offset += 1
    
    ax.set_xlim(-0.05, 10.05); ax.set_xticks(range(0, 11))
    ax.set_xlabel('Shake intensity reportada'); ax.set_yticks([])
    ax.set_title(f'Distribución de shake_intensity por barrio — {evento}', loc='left', fontsize=12)
    for s in ['top','right','left']:
        ax.spines[s].set_visible(False)
    plt.tight_layout()
    plt.show()

# Widget dropdown que regenera el ridge cuando cambia
evento_select = pn.widgets.Select(
    name='Evento sísmico',
    options=list(EVENTOS.keys()),
    value='Sismo principal (8/04)',
)

@pn.depends(evento_select.param.value)
def actualizar_ridge(evento):
    fig, ax = plt.subplots(figsize=(10, 8.5))
    t0, t1 = EVENTOS[evento]
    sub = df[(df['time'] >= t0) & (df['time'] <= t1)].dropna(subset=['shake_intensity'])
    
    x_grid = np.linspace(0, 10, 200)
    medians = sub.groupby('neighborhood')['shake_intensity'].median().sort_values(ascending=False)
    order = medians.index.tolist()
    if len(order) == 0:
        ax.text(0.5, 0.5, 'Sin datos en esta ventana', transform=ax.transAxes,
                ha='center', fontsize=14)
        return fig
    
    offset = 0; gap = 0.9
    colors_rg = plt.cm.plasma(np.linspace(0.15, 0.85, len(order)))
    
    for i, neigh in enumerate(order):
        vals = sub.loc[sub['neighborhood']==neigh, 'shake_intensity'].values
        if len(vals) < 5:
            ax.text(-0.15, offset+.25, neigh, ha='right', va='center', fontsize=9, color='gray')
            ax.text(10.15, offset+.25, f'n={len(vals)}', ha='left', va='center', fontsize=8, color='gray')
            offset += 1; continue
        try:
            kde = gaussian_kde(vals, bw_method=0.4)
            y = kde(x_grid); y = y / max(y.max(), 1e-6) * gap
        except Exception:
            offset += 1; continue
        ax.fill_between(x_grid, offset, offset+y, color=colors_rg[i], alpha=.78,
                        linewidth=.6, edgecolor='white')
        ax.plot(x_grid, offset+y, color='white', lw=.6)
        med = np.median(vals)
        ax.plot([med, med], [offset, offset+0.5], color='black', lw=1.2, alpha=.7)
        ax.text(-0.15, offset+.25, neigh, ha='right', va='center', fontsize=9)
        ax.text(10.15, offset+.25, f'n={len(vals):,}', ha='left', va='center', fontsize=8, color='gray')
        offset += 1
    
    ax.set_xlim(-0.05, 10.05); ax.set_xticks(range(0, 11))
    ax.set_xlabel('Shake intensity reportada'); ax.set_yticks([])
    ax.set_title(f'Distribución de shake_intensity por barrio — {evento}', loc='left', fontsize=12)
    for s in ['top','right','left']:
        ax.spines[s].set_visible(False)
    plt.tight_layout()
    plt.close(fig)
    return fig

pn.Column(evento_select, actualizar_ridge)
```

## Gráfico 10 — Reportes preocupantes por hora del día (dropdown por día)

```python
# Calendar heatmap con dropdown por día — ESCALA FIJA basada en el máximo global
df_ch = df.copy()
df_ch['fecha'] = df_ch['time'].dt.date
df_ch['hora'] = df_ch['time'].dt.hour
DIAS = sorted(df_ch['fecha'].unique())
DIAS_LABELS = [d.strftime('%d/%m (%a)') for d in DIAS]

# Calcular el máximo global a través de todos los días para usar como vmax fijo
def matriz_dia(fecha):
    sub = df_ch[df_ch['fecha'] == fecha]
    if len(sub) == 0:
        return None
    hm = sub.groupby('hora')[CATEGORIES].apply(lambda x: (x>=4).sum()).T
    hm.index = [CAT_ES[c] for c in hm.index]
    hm = hm.reindex(columns=range(24), fill_value=0)
    return hm

VMAX_GLOBAL = max(
    matriz_dia(d).values.max() if matriz_dia(d) is not None else 0
    for d in DIAS
)
print(f'VMAX global (escala fija): {VMAX_GLOBAL:,}')

def calendar_heatmap_dia(dia_label):
    idx = DIAS_LABELS.index(dia_label)
    fecha = DIAS[idx]
    fig, ax = plt.subplots(figsize=(13, 4))
    hm = matriz_dia(fecha)
    if hm is None or hm.sum().sum() == 0:
        ax.text(0.5, 0.5, f'Sin reportes preocupantes para {dia_label}',
                ha='center', va='center', transform=ax.transAxes, fontsize=14)
        plt.tight_layout()
        return fig

    sns.heatmap(hm, cmap='rocket_r', ax=ax, linewidths=.4, linecolor='white',
                cbar_kws={'label': 'Reportes con valor ≥ 4'},
                xticklabels=[f'{h:02d}h' for h in range(24)],
                vmin=0, vmax=VMAX_GLOBAL)  # ← ESCALA FIJA
    ax.set_title(f'Reportes preocupantes (≥4) por hora — {dia_label}\n(escala fija comparable entre días, máximo = {VMAX_GLOBAL:,})')
    ax.set_xlabel('Hora del día'); ax.set_ylabel('')
    plt.tight_layout()
    return fig

dia_select = pn.widgets.Select(
    name='Día',
    options=DIAS_LABELS,
    value=DIAS_LABELS[2] if len(DIAS_LABELS) > 2 else DIAS_LABELS[0],
)

@pn.depends(dia_select.param.value)
def actualizar_calendar(dia):
    fig = calendar_heatmap_dia(dia)
    plt.close(fig)
    return fig

pn.Column(dia_select, actualizar_calendar)
```

# Experimentos adicionales

```python
# Dependencias compartidas por los 3 gráficos de esta sección (ventana móvil de reportes
# por barrio, con slider temporal continuo).
gdf['centroid'] = gdf.geometry.centroid   # usado por el mapa de centroides y el cartograma Dorling

WINDOW_H = 3
slider_times = pd.date_range(df['time'].min(), df['time'].max(), freq='2h')
print(f'Frames del slider: {len(slider_times)}')

TH_SEV = sub_main.groupby('location')[DAMAGE].mean().mean(axis=1).median()
TH_SHAKE = sub_main.groupby('location')['shake_intensity'].mean().median()

def regimen_de(sev, shake):
    alta_sev = sev > TH_SEV
    alta_shake = shake > TH_SHAKE
    if alta_sev and alta_shake: return 'CONFIRMADO'
    if alta_sev and not alta_shake: return 'SORPRESA NEGATIVA'
    if not alta_sev and alta_shake: return 'SILENCIO SOSPECHOSO'
    return 'TRANQUILO'

def ventana_barrios(t_center, window_h=WINDOW_H):
    t0 = t_center - pd.Timedelta(hours=window_h/2)
    t1 = t_center + pd.Timedelta(hours=window_h/2)
    sub = df[(df['time'] >= t0) & (df['time'] < t1)].copy()
    if len(sub) == 0:
        return pd.DataFrame(columns=['location','severidad','shake','n_reportes','barrio','regimen'])
    sub['severidad_row'] = sub[DAMAGE].mean(axis=1)
    g = sub.groupby('location').agg(
        severidad=('severidad_row', 'mean'),
        shake=('shake_intensity', 'mean'),
        n_reportes=('time', 'size'),
    ).reset_index()
    g['barrio'] = g['location'].map(NEIGH)
    g['regimen'] = g.apply(lambda r: regimen_de(r['severidad'], r['shake']), axis=1)
    return g

frames = {pd.Timestamp(t): ventana_barrios(t) for t in slider_times}

# Tamaño del punto: exponente > 1 para que "explote" con reportes altos y sea
# prácticamente invisible (no solo chico) cuando hay muy pocos o ningún reporte.
N_MAX = max(g['n_reportes'].max() if len(g) else 0 for g in frames.values())
for t in frames:
    frames[t]['size_px'] = 55 * (frames[t]['n_reportes'] / N_MAX) ** 1.3
```

## Gráfico C — Versión geográfica: círculos en el centroide de cada barrio, con slider

Mismos datos que el Gráfico B del notebook experimental (reutiliza `frames`), pero posicionados en el

```python
base_polys = []
for _, row in gdf.iterrows():
    geom = row['geometry']
    if geom.geom_type == 'Polygon':
        xs, ys = geom.exterior.coords.xy
        base_polys.append({'x': list(xs), 'y': list(ys)})
base_map = hv.Polygons(base_polys).opts(fill_color='#f2f2f2', line_color='#999999', line_width=.6)

_centroides = gdf.set_index('location')['centroid']

def centroid_points_for(t):
    g = frames[t]
    g = g[g['n_reportes'] > 0].copy()   # sin reportes en la ventana → ni se dibuja el punto
    g['x'] = g['location'].map(lambda i: _centroides.get(i).x if i in _centroides.index else np.nan)
    g['y'] = g['location'].map(lambda i: _centroides.get(i).y if i in _centroides.index else np.nan)
    g = g.dropna(subset=['x', 'y'])
    return hv.Points(g, kdims=['x', 'y'], vdims=['barrio', 'n_reportes', 'severidad', 'size_px'])

hmap_centroids = hv.HoloMap({t: centroid_points_for(t) for t in slider_times}, kdims=[hv.Dimension('Tiempo')])

mapa_final = (base_map * hmap_centroids).opts(
    opts.Points(color='severidad', cmap='YlOrRd', size=hv.dim('size_px'),
                line_color='black', line_width=.6, tools=['hover'],
                colorbar=True, width=650, height=550,
                title='Reportes por barrio en el tiempo (radio = n reportes, color = severidad)'),
    opts.Polygons(width=650, height=550),
).opts(xaxis=None, yaxis=None)
mapa_final
```

## Gráfico D — Cartograma con slider temporal (dos variantes)

Estilo mapa de resultados electorales de EE.UU. — el tamaño de cada barrio "explota" o se achica según la
cantidad de reportes de la ventana móvil (3h), con slider temporal. Color = intensidad del sismo reportada
en esa misma ventana. Dos formas distintas de lograr el efecto de explosión:

- **Opción A — Cartograma de Dorling**: cada barrio se reemplaza por un círculo (área ∝ n reportes) que
  arranca en el centroide real y se separa de sus vecinos con un algoritmo de repulsión si se solapan.
- **Opción C — Explosión simple (escala in situ)**: la geometría real de cada barrio se escala directamente
  alrededor de su propio centroide (más simple, no corrige solapamientos, pero conserva la forma original).

### Opción A — Cartograma de Dorling

```python
# ============ OPCIÓN A — Cartograma de Dorling ============
import numpy as np
import holoviews as hv
from holoviews import opts
hv.extension('bokeh')

R_REF = np.sqrt(gdf.geometry.area.mean() / np.pi)   # ≈ 2660

def _dorling_pos(centros, radios, iteraciones=120):
    pos = centros.astype(float).copy()
    r = radios
    n = len(pos)
    for _ in range(iteraciones):
        for i in range(n):
            for j in range(i+1, n):
                d = pos[j] - pos[i]
                dist = np.hypot(*d)
                mind = r[i] + r[j]
                if 1e-9 < dist < mind:
                    push = (mind - dist) / 2
                    u = d / dist
                    pos[i] -= u * push
                    pos[j] += u * push
    return pos

def _circ(cx, cy, rad, k=44):
    a = np.linspace(0, 2*np.pi, k)
    return list(cx + rad*np.cos(a)), list(cy + rad*np.sin(a))

def polygons_dorling(t):
    g = ventana_barrios(t).set_index('location').reindex(range(1,20))
    n_rep = g['n_reportes'].fillna(0).values.astype(float)
    shake = g['shake'].fillna(0).values

    centros = np.array([[geom.x, geom.y] for geom in gdf['centroid']])
    # área ∝ n_reportes  →  radio ∝ sqrt(n). Normalizado al máximo GLOBAL
    # para que el tamaño sea comparable entre frames del slider.
    r = np.sqrt(np.maximum(n_rep, 1))
    r = r / np.sqrt(6237) * R_REF * 1.8     # 1.8 = factor de "explosión", subilo/bajalo a gusto

    pos = _dorling_pos(centros, r)
    data = []
    for i in range(len(gdf)):
        xs, ys = _circ(pos[i,0], pos[i,1], r[i])
        data.append({'x': xs, 'y': ys, 'shake': float(shake[i]),
                     'Barrio': gdf.iloc[i]['locName'], 'n_reportes': int(n_rep[i])})
    return hv.Polygons(data, vdims=['shake','Barrio','n_reportes'])

hmap_dorling = hv.HoloMap({t: polygons_dorling(t) for t in slider_times},
                          kdims=[hv.Dimension('Tiempo')])
dorling_final = hmap_dorling.opts(
    opts.Polygons(color='shake', cmap='YlOrRd', colorbar=True, clim=(0,10),
                  line_color='black', line_width=.6, tools=['hover'],
                  width=680, height=620, xaxis=None, yaxis=None,
                  title='Cartograma Dorling — área = n reportes, color = intensidad del sismo')
)
dorling_final
```

### Opción C — Explosión simple (escala in situ)

```python
# ============ OPCIÓN C — Explosión simple (escala in situ) ============
import numpy as np
import holoviews as hv
from holoviews import opts
from shapely.affinity import scale as shp_scale
hv.extension('bokeh')

def polygons_explosion(t):
    g = ventana_barrios(t).set_index('location').reindex(range(1,20))
    n_rep = g['n_reportes'].fillna(0).values.astype(float)
    shake = g['shake'].fillna(0).values
    # factor de escala 0.35x (barrio casi mudo) → 1.9x (barrio en pico), por sqrt(n)
    f = np.sqrt(np.maximum(n_rep, 1))
    f = 0.35 + 1.55 * (f / np.sqrt(6237))
    data = []
    for i, (_, row) in enumerate(gdf.iterrows()):
        geom = shp_scale(row['geometry'], xfact=f[i], yfact=f[i], origin='centroid')
        polys = geom.geoms if geom.geom_type == 'MultiPolygon' else [geom]
        for p in polys:
            xs, ys = p.exterior.coords.xy
            data.append({'x': list(xs), 'y': list(ys), 'shake': float(shake[i]),
                         'Barrio': row['locName'], 'n_reportes': int(n_rep[i])})
    return hv.Polygons(data, vdims=['shake','Barrio','n_reportes'])

hmap_expl = hv.HoloMap({t: polygons_explosion(t) for t in slider_times},
                       kdims=[hv.Dimension('Tiempo')])
explosion_final = hmap_expl.opts(
    opts.Polygons(color='shake', cmap='YlOrRd', colorbar=True, clim=(0,10),
                  line_color='black', line_width=.6, tools=['hover'],
                  width=680, height=620, xaxis=None, yaxis=None,
                  title='Explosión geográfica — tamaño = n reportes, color = intensidad')
)
explosion_final
```
